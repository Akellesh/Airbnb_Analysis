```python
def get_location_price_stats(collection):
    pipeline = [
        {"$group": {
            "_id": {"market": "$market", "suburb": "$suburb"},
            "avg_price": {"$avg": "$price"},
            "listing_count": {"$sum": 1}
        }},
        {"$sort": {"avg_price": -1}}
    ]
    results = list(collection.aggregate(pipeline))
    df = pd.DataFrame(results)
    df["market"] = df["_id"].apply(lambda x: x["market"])
    df["suburb"] = df["_id"].apply(lambda x: x["suburb"])
    df.drop(columns="_id", inplace=True)
    return df
```

```python
def get_neighbourhood_price_stats(collection):
    """Use MongoDB aggregation to get price stats per neighbourhood."""
    pipeline = [{"$group": {"_id": "$neighbourhood", "avg_price": {"$avg": "$price"}, "min_price": {"$min": "$price"},
                            "max_price": {"$max": "$price"}, "listing_count": {"$sum": 1}}}, {"$sort": {"avg_price": -1}}]
    results = list(collection.aggregate(pipeline))
    df_nps = pd.DataFrame(results)
    df_nps.rename(columns={"_id": "neighbourhood"}, inplace=True)
    return df_nps
```

```python
neighbourhood_stats = get_neighbourhood_price_stats(airbnb_collection)
```

```python
def get_monthly_review_counts(df):
    """Flatten nested review dates into a month-by-month count (seasonal proxy)."""
    all_dates = []
    for reviews in df["reviews"]:
        if isinstance(reviews, list):
            for review in reviews:
                if "date" in review:
                    all_dates.append(review["date"])

    dates_df = pd.DataFrame({"date": pd.to_datetime(all_dates)})
    dates_df["month"] = dates_df["date"].dt.month_name()
    monthly_counts = dates_df["month"].value_counts().reindex([
        "January", "February", "March", "April", "May", "June",
        "July", "August", "September", "October", "November", "December"
    ]).reset_index()
    monthly_counts.columns = ["month", "review_count"]
    return monthly_counts
```
